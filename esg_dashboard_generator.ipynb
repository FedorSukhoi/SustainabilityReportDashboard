# ESG Company Dashboard Generator

Generates one self-contained HTML dashboard per company from the analyzer's JSON outputs. The design uses an asymmetrical audit-oriented grid, restrained card boundaries, methodology context, detected certificates, accepted KPI evidence, and direct source-report links.

Run `esg_report_analyzer_v2.ipynb` first if `output/analysis/` is missing or stale.

```python
from pathlib import Path

ANALYSIS_DIR = Path("output/analysis")
DASHBOARD_DIR = Path("output/dashboards")

print(f"Analysis input:  {ANALYSIS_DIR.resolve()}")
print(f"Dashboard output: {DASHBOARD_DIR.resolve()}")
```

## Build all company dashboards

The rendering functions live in `dashboard_builder.py` so the same design can be regenerated from the notebook or the command line without duplicating templates.

```python
import importlib
import dashboard_builder

importlib.reload(dashboard_builder)
dashboard_files = dashboard_builder.build_all(ANALYSIS_DIR, DASHBOARD_DIR)

print(f"Generated {len(dashboard_files)} dashboards:")
for dashboard_file in dashboard_files:
    print(f" - {dashboard_file.resolve()}")
```

## Structural validation

Confirms that every generated document is self-contained and contains the expected auditability sections before visual review.

```python
required_markers = [
    "Sustainability overview",
    "Key sustainability areas",
    "Certificates & standards",
    "Sustainability evidence",
    "Evidence resolution",
]

for dashboard_file in dashboard_files:
    document = dashboard_file.read_text(encoding="utf-8")
    missing = [marker for marker in required_markers if marker not in document]
    assert not missing, f"{dashboard_file.name}: missing {missing}"
    assert "<script src=" not in document, f"{dashboard_file.name}: external script dependency found"
    print(f"OK  {dashboard_file.name:58} {dashboard_file.stat().st_size / 1024:7.1f} KB")
```
